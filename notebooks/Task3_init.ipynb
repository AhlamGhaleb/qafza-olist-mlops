```python
import numpy as np
import pandas as pd

# Load the feature matrix saved from Notebook 5
expected_X_test = pd.read_parquet(
    "data/artifacts/features/05_X_test.parquet"
)

# Get the feature names from the saved preprocessor
feature_names = preprocessor.get_feature_names_out()

# Convert Task 3 output to a DataFrame
actual_X_test = pd.DataFrame(
    processed_features,
    columns=feature_names
)

print("Expected shape:", expected_X_test.shape)
print("Actual shape:", actual_X_test.shape)

print(
    "Same columns:",
    list(expected_X_test.columns) == list(actual_X_test.columns)
)

print(
    "All values close:",
    np.allclose(
        expected_X_test.to_numpy(),
        actual_X_test.to_numpy(),
        rtol=1e-10,
        atol=1e-12
    )
)

max_difference = np.max(
    np.abs(
        expected_X_test.to_numpy()
        - actual_X_test.to_numpy()
    )
)

print("Maximum absolute difference:", max_difference)
```
